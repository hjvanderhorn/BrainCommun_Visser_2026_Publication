<h4>Short script for providing quality control metrics of sRNAbench mapping</h4>

```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_excel('/users/koen_phd/Documents/phd/biomarkers/mirna/rna_seq_quality_metrics.xlsx')
```

```python
data.head()
```

```python
interested_columns = ['raw reads', 'reads in analysis', 'UMI filtered reads', 'assignedUR', 'assignedRC', 'numberMatureMicroRNA', 'AdapterDimerPercent']

def provide_qc_metrics(dataframe, column):
    print(f'{column} median: {dataframe[column].median()}, IQR: {dataframe[column].quantile(0.25)} - {dataframe[column].quantile(0.75)}')

mtbi_data = data.iloc[0:54, :]
hc_data = data.iloc[54:, :]

for dataframe in [mtbi_data, hc_data]:
    for column in interested_columns:
        provide_qc_metrics(dataframe, column)
    print('--------------------------------')
```
